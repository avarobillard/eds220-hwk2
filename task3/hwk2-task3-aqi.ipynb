In this task you will use [Air Quality Index (AQI)](https://www.airnow.gov/aqi/aqi-basics/) data from the [US Environmental Protection Agency](https://www.epa.gov) to visualize the impact on the AQI of the 2017 [Thomas Fire](https://en.wikipedia.org/wiki/Thomas_Fire) in Santa Barbara County.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Save URLs to respective year variables
URL_2017 = 'https://aqs.epa.gov/aqsweb/airdata/daily_aqi_by_county_2017.zip'
URL_2018 = 'https://aqs.epa.gov/aqsweb/airdata/daily_aqi_by_county_2018.zip'

# Read in data 
aqi_17 = pd.read_csv(URL_2017)
aqi_18 = pd.read_csv(URL_2018)

# Concatenate the AQI 2017 and AQI 2018 data frames
aqi = pd.concat([aqi_17, aqi_18])

# Simplify column names
aqi.columns = (aqi.columns
                  .str.lower()
                  .str.replace(' ','_'))
               
# Filter to Santa Barbara county and drop unecessary columns
aqi_sb = aqi[aqi['county_name'] == 'Santa Barbara'].drop(columns = ['state_name', 'county_name', 'state_code', 'county_code'])
           
# Convert date column to timestamps 
aqi_sb.date = pd.to_datetime(aqi_sb.date)

# Set the date column as the index of the data frame
aqi_sb = aqi_sb.set_index('date')
               
# Calculate rolling average of AQI over 5 day windows and store in column 'five_day_average'
aqi_sb['five_day_average'] = aqi_sb['aqi'].rolling('5D').mean()
 
    
# Create plot
fig, ax = plt.subplots(figsize = (12,6))

# Daily AQI line
ax.plot(aqi_sb.index, aqi_sb['aqi'], color = '#7FADDC', linewidth = 1.2, label = "Daily AQI")

# Rolling average AQI line
ax.plot(aqi_sb.index, aqi_sb['five_day_average'], color = '#000080', linewidth = 1.5, label = "5-day rolling average AQI")

# Label axes and title
ax.set(xlabel = 'Date', ylabel = 'AQI value',
      title = 'U.S. Air Quality Index (AQI) Fluctuations in Santa Barbara, California (2017-2019)')

# Create vertical line to label Thomas Fire 
fire_date = pd.to_datetime('2017-12')

ax.vlines(x = fire_date, ymin = 30, ymax = 330, color = 'red', label = "Thomas Fire")

# Add legend
ax.legend()

plt.show()
```

In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()`method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as ouput

Store your answer in the `rolling_average` variable. You should have two columns in your series, `date` and the averages.
